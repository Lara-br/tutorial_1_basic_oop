### Points
We will start at the lowest level, this is the level of individual points of our fractals. This class utilises some python functionality outside the scope of this course and is included below.

```python
import random
from statistics import mean

class Point:
    def __init__(self, x: float, y: float):
        self.x = x
        self.y = y

    def __repr__(self):
        return f"Point({self.x}, {self.y})"

    @classmethod
    def random(self, x_min: float = 0, x_max: float = 1, y_min: float = 0, y_max: float = 1):
        return Point(random.uniform(x_min, x_max),
                     random.uniform(y_min, y_max))
    
    @classmethod
    def mean(self, points):
        x = mean(p.x for p in points)
        y = mean(p.y for p in points)
        return Point(x, y)
```

### Mandelbrot set

### Sierpinsky "Polygon"
This last fractal is a tad bit more complicated. A Sierpinsky triangle is the most famous example of this set of fractals. However, using the same algorithm, similar patterns can be made consisting of different shapes. All we need is a convex set of vertices to get started. A good starting point is {(-1, 0), (1, 0), (0.5, 1)}, this results in the most common Sierpinsky Triangle. 

There are many algorithms which can be used to generate the points of a Sierpinsky triangle. A common method is known as the chaos game (look it up on wikipedia). However we will be using a different (deterministic) algorithm with faster convergence. The steps of one iteration are specified below:

We first define a set of points added in the last iteration $\mathcal{P}$, in the first iteration this contains a singular starting point, this should be a point along one of the edges of the polygon. We also define $\mathcal{C}$ the set of corner points.
1. Take $p \in \mathcal{P}$ compute the middle of the line segment $pc$ where $c \in \mathcal{C}$, repeat this for each of the $n$ corner points. Add these to the fractal.
2. Repeat step 1 for each $p \in \mathcal{P}$.

This means that each iteration the amount of points added doubles, thus exponentially increasing the amount of points in the fractal.
